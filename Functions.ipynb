```python
def even_number(number):
    # Even number or not
    if number % 2 == 0:
        print("Even Number")
    else:
        print("Odd Number")

# Get user input
number = int(input("Enter a number: "))

# Call the function with the input
even_number(number)
```

```python

Employee Salary Calculation
20 % bonus if salary is 50k or more 10% bonus if salary is below 50k
2000 tax deduction if salary +bonus if greater than 60k
```

```python
# -Salary | Base Salary -Done
# -Calculate_bonus input salary => =50k 20% bonus , <50k 10% bonus
# =Total Salary = Base Salary+ bonus
# -> Calculate _tax - > total salary > 60k tax 2000,else 0 > return tax
# -> Total Salary, after tax , bonus, base salary


salary = int(input("Enter your salary: "))
def calculate_bonus(salary):
    if salary >= 50000:
        bonus = salary * 0.20
    else:
        bonus = salary * 0.10
    return bonus

bonus = calculate_bonus(salary)
total_salary = salary + bonus

def calculate_tax(total_salary):
    if total_salary > 60000:
        tax = 2000
    else:
        tax = 0
    return tax

tax = calculate_tax(total_salary)

total_salary_after_tax = total_salary - tax


# PRINT STATEMENT
print("Base Salary:", salary)
print("Bonus:", bonus)
print("Total Salary:", total_salary)
print("Tax:", tax)
print("Total Salary After Tax:", total_salary_after_tax)
```
